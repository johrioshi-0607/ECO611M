```python
f=open("sample.txt","w",encoding="utf-8")
f.write("India is my Country")
f.close()

file=open("sample.txt","r",encoding="utf-8")
text=file.read()

string=text.split(" ")
str1=""
for i in string:
    str1=str1+" "
    for ch in range(-1,-(1+len(i)),-1):
        str1=str1+i[ch]

print(str1)
        
```

```python
import pandas as pd
from pandas import DataFrame
f1=open("ques2.txt","w",encoding="utf-8")

data={'Name':['abc','def','ghi'],'Mark 1':[7,8,9],'Mark 2':[6,3,8]}
frame=pd.DataFrame(data)
frame

for col in frame.columns:
    values=frame[col].tolist()
    f1.write(f"{col}={values}\n")

f1.close()

f2=open("ques2.txt","r",encoding="utf-8")
print(f2.read())
```
