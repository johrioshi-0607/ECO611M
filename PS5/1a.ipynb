```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return np.cbrt(x) - np.exp(-x**2)

xm = None   # global variable

def root(x0, x1, n):
    global xl,xm
    xl.append(x0)
    xr.append(x1)
 
    xm = (x0 + x1)/2  
    
    if (x1 - x0)/2 <= 1e-6:
        return xm, n

    if f(xm) == 0:
        return xm, n

    if f(x0) * f(xm) < 0:
        n += 1
        return root(x0, xm, n)
    else:
        n += 1
        return root(xm, x1, n)


x0 = 0
x1 = 1
n = 0
xl=[]
xr=[]
xm, n = root(x0, x1, n)

print("root =", xm)
print("n =", n)
print("f(root) =", f(xm))

plt.plot(xl, marker="o", label="Left limits (xl)")
plt.plot(xr, marker="s", label="Right limits (xr)")

plt.xlabel("Iteration k")
plt.ylabel("x value")
plt.title("Bisection Trajectories of xl and xr")
plt.grid(True)
plt.legend()
plt.show()
```
