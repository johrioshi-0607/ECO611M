```python
# Make an orthogonal matrix
import numpy as np

def A_matrix(n):
    basis = []

    while len(basis) < n:
        v = np.random.randn(n)

        for q in basis:
            v = v - np.dot(q, v) * q

        norm = np.linalg.norm(v)
        if norm > 1e-10:
            basis.append(v / norm)

    B = np.column_stack(basis)

    # Correct orthonormality check
    print(B.T @ B)

    return B

A_matrix(5)

```
