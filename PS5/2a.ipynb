```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.cbrt(x) - np.exp(-x**2)
def f1(x):
    return (1/3)*x**(-2/3)+2*x*np.exp(-x**2)

def root(x0,n):
    global xn
    xn=x0-(f(x0)/f1(x0))
    xl.append(xn)
    if np.abs(xn-x0)<1e-6:
        return xn,n
    if f(xn)==0:
        return xn,n
    else:
        return root(xn,n+1)


x0=0.1
xl=[x0]
xn=0
n=0

xn,n=root(x0,n)
print("root=",xn)
print("n=",n)
print("f(root)=",f(xn))

plt.plot(xl)
plt.show()
```
