```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 8 - 12*x + 86*x**2 - 121*x**3 + 60*x**4 - 10*x**5
def f1(x):
    return -12 + 172*x - 363*x**2 + 240*x**3 - 50*x**4

def root(x0,n):
    global xn
    xn=x0-(f(x0)/f1(x0))
    xl.append(xn)
    if np.abs(xn-x0)<1e-6:
        return xn,n
    if f(xn)==0:
        return xn,n
    else:
        return root(xn,n+1)


x0=0.1
xl=[x0]
xn=0
n=0

xn,n=root(x0,n)
print("root=",xn)
print("n=",n)
print("f(root)=",f(xn))

plt.plot(xl)
plt.show()
```
