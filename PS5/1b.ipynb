```python
import numpy as np
import matplotlib.pyplot as plt
def f2(x):
    return x**4-3*x**3+4*x**2+5*x-2

def root2(x0,x1,n):
    global xl,xm
    xl.append(x0)
    xr.append(x1)
    global xm
    xm=(x1+x0)/2
    
    if (x1-x0)/2<=1e-6:
        return xm,n

    if f2(xm)==0:
        return xm,n
        
    if f2(x0)*f2(xm)<0:
        return root2(x0,xm,n+1)
    else:
        return root2(xm,x1,n+1)

x0=0
x1=1
n=0
xl=[]
xr=[]
xm,n=root2(x0,x1,n)
print("root =", xm)
print("n =", n)
print("f(root) =", f2(xm))

plt.plot(xl, marker="o", label="Left limits (xl)")
plt.plot(xr, marker="s", label="Right limits (xr)")

plt.xlabel("Iteration k")
plt.ylabel("x value")
plt.title("Bisection Trajectories of xl and xr")
plt.grid(True)
plt.legend()
plt.show()
```
